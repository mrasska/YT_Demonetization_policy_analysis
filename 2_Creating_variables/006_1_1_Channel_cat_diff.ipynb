```python
import pandas as pd
import numpy as np
```

```python
db51 = pd.read_csv('channel_cat.csv', sep=';')

db80 = pd.read_csv('channel_cat_80.csv', sep=';')
```

```python
comp_db = pd.merge(db51, db80, how='inner', left_on='channel', right_on='channel')
comp_db
```

```python
conditions = [(comp_db['main_cat_x']==comp_db['main_cat_y']), 
              (comp_db['main_cat_x']!=comp_db['main_cat_y'])]

values = [True, False]

comp_db['same'] = np.select(conditions, values)

comp_db
```
