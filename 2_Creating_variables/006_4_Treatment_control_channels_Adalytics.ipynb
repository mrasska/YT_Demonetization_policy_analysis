## Treatment and control groups for selected channels

The examined intervention targeted specific content: "videos that are perceived to be hateful or inflammatory” and “content that is harassing or attacking people based on their race, religion, gender or similar categories".
To estimate the effect on creators’ content supply, we categorised channels based on the topic of their published videos before the implementation of the moderation. The categorisation relied on keyword matching. A total of 96 words and expressions based on this condition  was generated. 
The keyword matching was done on the title, description and tags for videos published before the March 2017’s moderation update (n videos = 2,742,395).

## Title, description and tags analysis

```python
bad_words=[]

path = 'adalytics_cleaned_2026.txt'

with open(path, 'r') as fp:
    for line in fp:
        x = line[:-1]
        # add current item to the list
        bad_words.append(x)

bad_words
```

```python
len(bad_words)
```

```python
bad_words = re.compile('|'.join(bad_words))   
mask = ((db_before['title'].str.contains(bad_words))|(db_before['description'].str.contains(bad_words))|(db_before['tags'].str.contains(bad_words)))
db_before['bad_words']=mask
db_before['bad_words']=db_before['bad_words'].fillna('None')
```

```python
db_before
```

```python
db_before['bad_words'].value_counts()
```

```python
db_channels=db_before.groupby(['channel','bad_words'], as_index=False).size()
db_channels
```

```python
db_cha_bad=pd.crosstab(db_before['channel'],db_before['bad_words'], margins=True)
db_cha_bad
```

```python
db_cha_bad['percentage_true']=round((db_cha_bad[True]/db_cha_bad['All'])*100)

conditions=[(db_cha_bad['percentage_true']<=0),
            (db_cha_bad['percentage_true']>0)]
values=['Control','Treatment']

db_cha_bad['type_channel_0']=np.select(conditions, values)

db_cha_bad
```

```python
conditions=[(db_cha_bad['percentage_true']<=10),
            (db_cha_bad['percentage_true']>10)]
values=['Control','Treatment']

db_cha_bad['type_channel_10']=np.select(conditions, values)

db_cha_bad
```

```python
conditions=[(db_cha_bad['percentage_true']<=25),
            (db_cha_bad['percentage_true']>25)]
values=['Control','Treatment']

db_cha_bad['type_channel_25']=np.select(conditions, values)

db_cha_bad
```
