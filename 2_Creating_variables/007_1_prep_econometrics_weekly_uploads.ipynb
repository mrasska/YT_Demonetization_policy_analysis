```python
import pandas as pd
import numpy as np
```

## Regrouping data per weeks

```python
db_video=pd.read_csv('weekly_uploads_cat_completed.csv', sep=';')
db_video.info()
```

```python
# videos published after 1st content moderation update
condition=[(db_video["week"]>="2017-12"),
          (db_video["week"]<"2017-12")]
value=[1, 0]
db_video["first_moderation_update"]=np.select(condition, value)
```

```python
db_video
```

```python
db_video['week']=db_video['week'].replace('2017-00', '2016-52')
```

```python
db_video
```

```python
db_weekly=db_video.groupby(['channel', 'week', 
                           'first_moderation_update'],
                           as_index=False, 
                           dropna=False)['id_0'].sum(min_count=1)
db_weekly
```

```python
db_weekly[db_weekly['week']=='2016-52']
```

### Coding dummies

```python
condition=[(db_weekly['type_channel_0']=='Control'),
          (db_weekly['type_channel_0']=='Treatment')]
value=[0,1]
db_weekly['treated_ada_0']=np.select(condition, value)

condition=[(db_weekly['type_channel_10']=='Control'),
          (db_weekly['type_channel_10']=='Treatment')]
value=[0,1]
db_weekly['treated_ada_10']=np.select(condition, value)

condition=[(db_weekly['type_channel_25']=='Control'),
          (db_weekly['type_channel_25']=='Treatment')]
value=[0,1]
db_weekly['treated_ada_25']=np.select(condition, value)
```
